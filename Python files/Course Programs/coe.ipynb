```python
# Logical Operations using Sets 
album_set1 = set(["Thriller", 'AC/DC', 'Back in Black'])
album_set2 = set([ "AC/DC", "Back in Black", "The Dark Side of the Moon"])
album_set1, album_set2
```

```python
# Find the intersections

intersection = album_set1 & album_set2 # The and (&) finds what is common between the two 
intersection
```

```python
# Find the difference in set1 but not set2

album_set1.difference(album_set2)  
album_set2.difference(album_set1)  
```

```python
# Use intersection method to find the intersection of album_set1 and album_set2

album_set1.intersection(album_set2)  
```

```python
# Find the union of two sets

album_set1.union(album_set2)
```

```python
# Check if superset

set(album_set1).issuperset(album_set2)   
set(album_set2).issubset(album_set1)  
```

```python
# Comparison and branching , Loops


 # For loop example

dates = [1982,1980,1973]
N = len(dates)

for i in range(N):
    print(dates[i])     
```

```python
# Use for loop to change the elements in list

squares = ['red', 'yellow', 'green', 'purple', 'blue']

for i in range(0, 5):
    print("Before square ", i, 'is',  squares[i])
    squares[i] = 'white'
    print("After square ", i, 'is',  squares[i])
```

```python
# Loop through the list and iterate on both index and element value

squares=['red', 'yellow', 'green', 'purple', 'blue']

for i, square in enumerate(squares):
    print(i, square)
```

```python
# While Loop Example

dates = [1982, 1980, 1973, 2000]

i = 0
year = dates[0] # 1982 , 1980

while(year != 1973):    
    print(year)
    i = i + 1
    year = dates[i]
    

print("It took ", i ,"repetitions to get out of loop.")
```

```python
count = 0
while count < 10:
    count += 1
    if count == 3:
        continue  # skip printing 3
    if count == 8:
        break     # stop the loop when count is 8
    print(count)
    
```
