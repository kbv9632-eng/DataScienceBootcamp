```python
# Logical Operations using Sets 
album_set1 = set(["Thriller", 'AC/DC', 'Back in Black'])
album_set2 = set([ "AC/DC", "Back in Black", "The Dark Side of the Moon"])
album_set1, album_set2
```

```python
# Find the intersections

intersection = album_set1 & album_set2 # The and (&) finds what is common between the two 
intersection
```

```python
# Find the difference in set1 but not set2

album_set1.difference(album_set2)  
album_set2.difference(album_set1)  
```

```python
# Use intersection method to find the intersection of album_set1 and album_set2

album_set1.intersection(album_set2)  
```

```python
# Find the union of two sets

album_set1.union(album_set2)
```

```python
# Check if superset

set(album_set1).issuperset(album_set2)   
set(album_set2).issubset(album_set1)  
```

```python
# Use for loop to change the elements in list

squares = ['red', 'yellow', 'green', 'purple', 'blue']

for i in range(0, 5):
    print("Before square ", i, 'is',  squares[i])
    squares[i] = 'white'
    print("After square ", i, 'is',  squares[i])
```

```python
# Loop through the list and iterate on both index and element value

squares=['red', 'yellow', 'green', 'purple', 'blue']

for i, square in enumerate(squares):
    print(i, square)
```

```python
# While Loop Example

dates = [1982, 1980, 1973, 2000]

i = 0
year = dates[0] # 1982 , 1980

while(year != 1973):    
    print(year)
    i = i + 1
    year = dates[i]
    

print("It took ", i ,"repetitions to get out of loop.")
```

```python
# Decorators 

def before_after(func):

    def wrapper():
       print('Before Function ')
       func()
       print('After Function')

    return wrapper



@before_after
def greet():
    print("Hello bro")



greet()
```

```python
def changecase(func):
  def myinner():
    return func().upper()
  return myinner

@changecase
def myfunction():
  return "Hello Sally"

print(myfunction())
```

```python
def changecase(func):
  def myinner():
    return func() + '!!!'
  return myinner

@changecase
def myfunction():
  return "Hello Sally"

print(myfunction())
```

```python
def add_exclamation(func):
    def inner(name):
        return func(name)+ ' !'
    return inner


@add_exclamation
def greet(name):
    return "Hello " + name



print(greet("Kiran"))
```

```python
def log_arguments(func):
    def inner_func(*numbers):
        print('Args: ', *numbers)
        return func(*numbers)
    return inner_func


@log_arguments
def add(a, b, c):
    return a + b + c


print(add(1,2,3))
```

```python

def log_arguments(func):
    def inner_func(*numbers):
        print(f'Args are:- ', *numbers)
        return func(*numbers)
    return inner_func




@log_arguments
def add(a, b, c):
    return a + b + c

print(add(1,1,1))

```

```python
def show_info(func):
    def inner_func(*pos_info,**key_info):
        print('The arguments are : ',*pos_info, *key_info)
        return func(*pos_info,**key_info)
    return inner_func




@show_info
def introduce(name, age):
    return f"My name is {name} and I am {age}"


print(introduce("Kiran", age=24))
```

```python
def check_access(func):
    def inner_func(user_role):
      if user_role.lower() == 'admin':
         
         return func(user_role)    
      else:
         return 'Access Denied'
      
    return inner_func



@check_access
def delete_db(role):
    return 'DB Deleted'




print(delete_db('Admin'))
```

```python

def log_call(func):
    def inner_func(*args,**kwargs):
        print('The Arguments are: ',args, kwargs)
        return func(*args,**kwargs)
    return inner_func
        




@log_call
def create_user(name, age, city="Bangalore"):
    return f"{name}, {age}, {city}"



create_user("Kiran", 24, city="Bangalore")
```

```python
def timer(func):
    def inner_func():
        return func()
    return inner_func


@timer
def do_work():
    total = 0

    for i in range(1000000):
        total += i

    return total



result = do_work()
print(result)
```
