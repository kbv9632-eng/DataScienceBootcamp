```python
def multiply(a, b, c):
    return a * b * c


def wrapper(func, *args):
    print("Starting")
    
    result = func(*args)
    
    print("Finished")
    
    return result


x = wrapper(multiply, 2, 3, 4)

print(x)
```

```python
# this is for known nos of args 
def add(a, b):
    return a + b

def wrapper(func,a,b):
    result = func(a,b)
    return result

wrapper(add,1,2)

# but in case of unknown nos of args you can always use *args - Packs args while collecting into tuple
def add(a,b,c):
    return a+b+c

def wrapper(func,*num):
    result = func(*num)
    return result


sample = wrapper(add,1,2,3)
print(sample)
```

```python
# Sample Question with *Args 

def multiply(*nums):
    result = 1
    for num in nums:
        result = num * result

    return result 




print(multiply(2, 3, 4))
print(multiply(2, 3, 4, 5))
```

```python
def largest(*nums):
    max = nums[0] # make sure that everytime you deal with a starting var intialize the value to first val of the data structure the data is stored in 
    for num in nums:
        if num >= max:
            max = num

    return max

largest(4, 9, 2, 7)
largest(100, 25, 300, 12, 50)
```
