```python
# Q2 
def analyze_numbers(numbers):
    even_count = 0
    odd_count = 0 

    for num in numbers:
        if num %2 == 0:
            even_count += 1
        else:
            odd_count += 1
    return even_count , odd_count


analyze_numbers([10, 3, 7, 8, 12, 5])
```

```python
# Day 2 
# Q1 

def list_cal(list):
    min_num = list[0]
    max_num = list[0]
    sum = 0
    for num in list:
        if num > max_num:
            max_num = num
        if num < min_num:
            min_num = num 
        sum += num 
    return max_num,min_num,sum



list_cal([10, 25, 3, 40, 15])
    
```
