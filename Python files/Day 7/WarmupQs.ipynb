```python
# Question 1 

students = [
    {
        "student_id": "101",
        "name": "Kiran",
        "age": 22,
        "gender": "Male",
        "student_class": "BSc DS"
    },
    {
        "student_id": "102",
        "name": "Alice",
        "age": 20,
        "gender": "Female",
        "student_class": "BSc AI"
    },
    {
        "student_id": "103",
        "name": "John",
        "age": 19,
        "gender": "Male",
        "student_class": "BSc DS"
    },
    {
        "student_id": "104",
        "name": "Sarah",
        "age": 19,
        "gender": "Female",
        "student_class": "BSc AI"
    }
]



def count_students_above_age():
    count = 0 
    user_inpt = int(input("Enter Age:- "))
    if user_inpt == 0 :
        return  
    for student in students:
        if student['age'] > user_inpt:
           count += 1
    if count > 0:
        print('There are {0} students older than {1}'.format(count,user_inpt))
    elif count == 0:
        print('No students found above the given age.')
       


count_students_above_age()
```

```python
 # Warmup Qs

#Q1

def count_male_students():
    if len(students) == 0:
        print("Database is empty!")
        return 
    else:
        count = 0
        for student in students:            
            if student['gender'] == 'Male':
                count += 1

        print('Total Male Students: ', count)


count_male_students()

```

```python
# Q2
def find_students_by_class():
    if not students:
        print("Database is empty!")
        return 
    user_inp = input('Enter Class : ')
    found = False
    
    for student in students:
        if user_inp.strip().lower() == student["student_class"].strip().lower() :
            print(student)
            found = True
    if not found:
        print('No students found in this class.')


find_students_by_class()
```

```python
# Q3 
def count_students_younger_than():
    if not students:
        print("Database is empty!")
        return
    
    user_age = int(input('Enter Age: '))
    count = 0 
    

    for student in students:
        if student['age'] < user_age :
            count += 1
            
    
    print('Total Students Younger Than {0}: {1}'.format(user_age,count))
   

count_students_younger_than()
```
