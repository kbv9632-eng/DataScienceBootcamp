```python
# Actual Solution 
def data_clean(customers):

    cleaned_customers = []

    for customer in customers:

        # Clean the name
        name = customer["name"].strip().capitalize()

        # Clean the age
        try:
            age = int(customer["age"])
        except (ValueError, TypeError):
            age = None

        # Create ONE dictionary for this customer
        cleaned_customer = {
            "name": name,
            "age": age
        }

        # Add the entire customer to the list
        cleaned_customers.append(cleaned_customer)

    return cleaned_customers
```
