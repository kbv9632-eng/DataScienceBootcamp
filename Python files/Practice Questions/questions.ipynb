```python
# Q1 
def transaction(*transactions):
    if len(transactions) !=0:
        sum_transactions = 0 
        avg_transactions = 0 
        largest_transactions = transactions[0]
        smallest_transactions = transactions[0]
        

        for transaction in transactions:
            sum_transactions = sum_transactions + transaction # to calculate the sum 
            

            if transaction > largest_transactions:
                largest_transactions = transaction

            if transaction < smallest_transactions:
                smallest_transactions = transaction

        avg_transactions = sum_transactions / len(transactions)

        return sum_transactions,avg_transactions,largest_transactions,smallest_transactions
    else :
        print("Please provide arguments to the function")
    
```
